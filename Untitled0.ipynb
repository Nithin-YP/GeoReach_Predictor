Normalize / standardize features

Standardize income, literacy, etc., to a common scale (0–1).

Engineer “interest_score” or “demand alignment”

These datasets often lack direct variables for “interest in product category” or “consumption preferences.”

### Subtask:
Further process the merged `district_population` and `unemployment` data. This may involve handling missing values, aggregating unemployment data to a suitable level (e.g., state or district and possibly a specific time period), and cleaning/converting data types (e.g., Population and Growth columns).

Examine the merged_df DataFrame to understand the extent of missing values and decide on an aggregation strategy. Then, aggregate the unemployment data by 'State' and 'Area' and clean 'Population' and 'Growth' columns by removing non-numeric characters and converting to numeric types.



```python
print("\nMissing values in merged_df before cleaning and aggregation:")
print(merged_df.isnull().sum())

# Aggregate unemployment data by 'State' and 'Area'
unemployment_agg = merged_df.groupby(['State', 'Area'])[['Estimated Unemployment Rate (%)', 'Estimated Employed', 'Estimated Labour Participation Rate (%)']].mean().reset_index()

# Merge the aggregated unemployment data back to the main dataframe
merged_df = merged_df.drop(['Region', 'Date', 'Frequency', 'Estimated Unemployment Rate (%)', 'Estimated Employed', 'Estimated Labour Participation Rate (%)', 'Area'], axis=1)
merged_df = pd.merge(merged_df, unemployment_agg, on='State', how='left')


# Clean and convert 'Population' column
merged_df['Population'] = merged_df['Population'].str.replace(',', '', regex=False).astype(float)

# Clean and convert 'Growth' column
merged_df['Growth'] = merged_df['Growth'].str.replace('%', '', regex=False).astype(float)

# Display the updated merged dataframe info and missing values
print("\nMerged DataFrame after cleaning and aggregation:")
display(merged_df.head())
merged_df.info()
print("\nMissing values in merged_df after cleaning and aggregation:")
print(merged_df.isnull().sum())
```

**Reasoning**:
Handle the remaining missing values in the merged_df DataFrame by imputing numerical values with the mean and categorical values with the mode.



```python
# Impute missing numerical values with the mean
numerical_cols = ['Estimated Unemployment Rate (%)', 'Estimated Employed', 'Estimated Labour Participation Rate (%)']
for col in numerical_cols:
    merged_df[col].fillna(merged_df[col].mean(), inplace=True)

# Impute missing categorical values with the mode or 'Unknown'
categorical_cols = ['Area']
for col in categorical_cols:
    merged_df[col].fillna(merged_df[col].mode()[0] if not merged_df[col].mode().empty else 'Unknown', inplace=True)


# Display missing values after imputation
print("\nMissing values in merged_df after imputation:")
print(merged_df.isnull().sum())
```

## Process employment data

### Subtask:
Clean and process the `employment_data`. This might involve aggregating employment status, salary, and industry information to a geographical level to be merged with other data.


**Reasoning**:
Inspect the employment_data DataFrame to understand its structure, columns, and data types, paying close attention to columns that could be used for geographical linking or contain relevant employment information (e.g., 'Location', 'Status', 'Monthly Salary (INR)', 'Industry'). Check for and handle any missing values in the relevant columns.



```python
print("Employment Data:")
display(employment_data.head())
employment_data.info()

print("\nMissing values in employment_data before cleaning:")
print(employment_data.isnull().sum())

# Handle missing values in relevant columns.
# 'Years of Experience' and 'Monthly Salary (INR)' have missing values, impute with median.
# 'Location' and other categorical columns do not have missing values based on info(), but we'll verify unique values.
employment_data['Years of Experience'].fillna(employment_data['Years of Experience'].median(), inplace=True)
employment_data['Monthly Salary (INR)'].fillna(employment_data['Monthly Salary (INR)'].median(), inplace=True)

print("\nMissing values in employment_data after imputation:")
print(employment_data.isnull().sum())

print("\nUnique values in 'Location' column:")
print(employment_data['Location'].value_counts())

print("\nUnique values in 'Status' column:")
print(employment_data['Status'].value_counts())

print("\nUnique values in 'Industry' column:")
print(employment_data['Industry'].value_counts())
```

**Reasoning**:
Based on the inspection, the 'Location' column contains a mix of specific cities (Mumbai, Hyderabad, Delhi, Bangalore) and broader categories (Urban, Suburban, Rural). A direct merge with state/district level data using 'Location' as is will be challenging. The best approach is to aggregate the employment features by the unique values present in the 'Location' column. We will calculate the average monthly salary, employment rate (proportion of 'Employed' status), and perhaps the most frequent industry for each unique location.



```python
# Calculate Employment Rate per location
employment_data['is_employed'] = (employment_data['Status'] == 'Employed').astype(int)
employment_rate_agg = employment_data.groupby('Location')['is_employed'].mean().reset_index()
employment_rate_agg.rename(columns={'is_employed': 'Employment_Rate'}, inplace=True)

# Calculate Average Monthly Salary per location
salary_agg = employment_data.groupby('Location')['Monthly Salary (INR)'].mean().reset_index()
salary_agg.rename(columns={'Monthly Salary (INR)': 'Avg_Monthly_Salary'}, inplace=True)

# Determine the most frequent Industry per location (Mode)
industry_mode_agg = employment_data.groupby('Location')['Industry'].agg(lambda x: x.mode()[0] if not x.mode().empty else 'Unknown').reset_index()
industry_mode_agg.rename(columns={'Industry': 'Most_Frequent_Industry'}, inplace=True)


# Merge the aggregated dataframes
employment_agg_df = pd.merge(employment_rate_agg, salary_agg, on='Location')
employment_agg_df = pd.merge(employment_agg_df, industry_mode_agg, on='Location')

# Display the aggregated employment dataframe
print("\nAggregated Employment Data by Location:")
display(employment_agg_df.head())
employment_agg_df.info()
print("\nMissing values in aggregated employment_data:")
print(employment_agg_df.isnull().sum())
```
